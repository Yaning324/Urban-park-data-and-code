## Panel b Overall Top-5 co-dominance chord diagram

Source: `RAM_Tag_Merge_Coverage_and_Semantic_Summary.ipynb`, cell 45. The existing Top-5 table supplies the park identifiers and tags. The category codes follow Appendix B.2.


```python
import pandas as pd
import holoviews as hv
from holoviews import opts
from itertools import combinations
import os
import datetime
import collections

# Enable the Bokeh backend
hv.extension('bokeh')

output_directory = str(OUTPUT_DIR)
os.makedirs(output_directory, exist_ok=True)
current_timestamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
from IPython.display import display
df_top5_merged = pd.read_excel(INPUT_DIR / "每个公园top5和重分类.xlsx", sheet_name="各个公园总")
df_top5_merged = df_top5_merged[["代号", "词语"]].rename(columns={"代号": "ParkID", "词语": "Tag"})
# Appendix B.2 classifications for the 29 tags in the park-level Top-5 table.
tags_by_category = {
    "G1": "flower grass plant tree",
    "G2": "autumn fall hill hillside lake pond sky stone water",
    "G3": "bench building floor path pavement pillar playground sign",
    "G4": "city field garden park temple",
    "G5": "person",
    "G7": "green lush",
}
category_by_tag = {tag: group for group, tags in tags_by_category.items() for tag in tags.split()}
tag_keys = df_top5_merged["Tag"].str.strip().str.casefold()
df_top5_merged["Category"] = tag_keys.map(category_by_tag)
park_sizes = df_top5_merged.groupby("ParkID").size()
if len(park_sizes) != 31 or not park_sizes.eq(5).all() or df_top5_merged["Category"].isna().any():
    raise ValueError("Expected five classified Top-5 tags for each of the 31 study parks.")
```

```python
# Skip plotting if the dataset is empty
if not df_top5_merged.empty:
    # 3. Colour mapping: reuse the updated bar chart colours
    COLOR_MAP_HEX_NEW = {
        "G1": "#6BA368",
        "G2": "#5B8DB8",
        "G3": "#A0A4A8",
        "G4": "#D8A47F",
        "G5": "#C06C84",
        "G6": "#7A6C9D",
        "G7": "#8AA6A3",
        "G8": "#C9C9A3",
        "Uncategorized": "#BBBBBB"
    }

    # Helpers to convert hex colours to RGB strings and blend RGB colours
    def hex_to_rgb_string(hex_color):
        hex_color = hex_color.lstrip('#')
        lv = len(hex_color)
        rgb = tuple(int(hex_color[i:i + lv // 3], 16) for i in range(0, lv, lv // 3))
        return f'rgb({rgb[0]},{rgb[1]},{rgb[2]})'

    def rgb_string_to_tuple(rgb_str):
        parts = rgb_str.replace('rgb(', '').replace(')', '').split(',')
        return tuple(int(p.strip()) for p in parts)

    def rgb_tuple_to_string(rgb_tuple):
        return f'rgb({rgb_tuple[0]},{rgb_tuple[1]},{rgb_tuple[2]})'

    def blend_colors(color1_rgb_str, color2_rgb_str):
        rgb1 = rgb_string_to_tuple(color1_rgb_str)
        rgb2 = rgb_string_to_tuple(color2_rgb_str)
        blended_r = (rgb1[0] + rgb2[0]) // 2
        blended_g = (rgb1[1] + rgb2[1]) // 2
        blended_b = (rgb1[2] + rgb2[2]) // 2
        return rgb_tuple_to_string((blended_r, blended_g, blended_b))

    # 4. Prepare the node data
    node_df = df_top5_merged[['Tag', 'Category']].drop_duplicates().rename(columns={'Tag': 'name', 'Category': 'group'})
    node_df['color'] = node_df['group'].map(lambda x: hex_to_rgb_string(COLOR_MAP_HEX_NEW.get(x, '#BBBBBB')))
    node_df = node_df.set_index('name')

    # 5. Prepare the edge data from the co-occurrence matrix
    co_occurrence_counts = collections.defaultdict(int)
    grouped_tags = df_top5_merged.groupby('ParkID')['Tag'].apply(lambda x: list(set(x)))

    for tags in grouped_tags:
        if len(tags) < 2: continue
        for t1, t2 in combinations(tags, 2):
            if t1 < t2:
                co_occurrence_counts[(t1, t2)] += 1
            else:
                co_occurrence_counts[(t2, t1)] += 1

    edge_data = [{'source': s, 'target': t, 'value': v} for (s, t), v in co_occurrence_counts.items() if v > 0]
    df_edges = pd.DataFrame(edge_data)

    # 6. Filter nodes and blend the colours of connected nodes for each edge
    if not df_edges.empty:
        involved_tags = pd.Series(df_edges['source'].tolist() + df_edges['target'].tolist()).unique()
        node_df_filtered = node_df.loc[involved_tags].reset_index()
        node_df_filtered = node_df_filtered.sort_values(by=['group', 'name']).reset_index(drop=True)

        tag_colors_map = node_df_filtered.set_index('name')['color'].to_dict()

        blended_edge_colors = []
        for _, row in df_edges.iterrows():
            c1 = tag_colors_map.get(row['source'])
            c2 = tag_colors_map.get(row['target'])
            if c1 and c2:
                blended_edge_colors.append(blend_colors(c1, c2))
            else:
                blended_edge_colors.append('rgb(128,128,128)')
        df_edges['blended_color'] = blended_edge_colors

        # 7. Draw the chord diagram
        nodes = hv.Dataset(node_df_filtered, 'name')
        chord = hv.Chord((df_edges, nodes)).opts(
            opts.Chord(
                labels='name',
                node_color='color',
                width=800, height=800, aspect='square',
                node_size=8,
                node_line_width=2, # Use a thicker outer outline to define the boundary
                node_line_color='white', # Set edge outlines to white to separate the segments clearly
                padding=0.1,
                edge_line_width=0.5,
                edge_color='blended_color',
                label_text_font_size='10pt',
                title="Top 5 Tags Co-occurrence (New 8-Category System)",
                hover_tooltips=[('Tag', '@name'), ('Category', '@group'), ('Connections', '@value')]
            )
        )

        # 8. Save and display the diagram
        output_html_path = os.path.join(output_directory, f'new_category_top5_chord_diagram_{current_timestamp}.html')
        hv.save(chord, output_html_path, backend='bokeh')

        print(f"基于新分类的 Top 5 词语共现弦图已保存至: {output_html_path}")
        display(chord)
    else:
        print("警告: 没有找到任何共现关系数据，无法生成弦图。")
```
